```python
def get_sqrt_lst(lst):
  new_lst = []
  for val in lst:
    new_lst.append(math.sqrt(val))
  return new_lst
```

```python
def get_max_pages(code):
  data =  {
      "subject": None,
      "organization": {"name":None,"inn":None,"ogrn":None,"place":None,"type":None,"studyType":None},
      "person": {"lastname":None,"firstname":None,"patronymic":None,"place":None,"egrip":None,"inn":None},
      "certificate": {"registerNumber":None,"series":None,"number":None, "startDate":{"startDate":None,"endDate":None},"endDate":{"startDate":None,"endDate":None},"status":12},
      "extended": True,
      "supplements": {"status":None,"code":code,"name":None,"level":None},
      "searchType": 1,
      "page": 1,
      "per_page": 100
  }

  res = requests.post('https://isga.obrnadzor.gov.ru/api/spa/register/getRegisterTable', json=data)
  pages = res.json()['last_page']

  return pages
```

#### Пишем функцию, которая будет получать информацию про конкретную образовательную организацию

```python
res = requests.post('https://isga.obrnadzor.gov.ru/api/spa/register/getRegisterTable', json=data)
```

```python
def get_orgs_list(code, page):
    data = {
        "subject": None,
        "organization": {"name":None,"inn":None,"ogrn":None,"place":None,"type":None,"studyType":None},
        "person": {"lastname":None,"firstname":None,"patronymic":None,"place":None,"egrip":None,"inn":None},
        "certificate": {"registerNumber":None,"series":None,"number":None, "startDate":{"startDate":None,"endDate":None},"endDate":{"startDate":None,"endDate":None},"status":12},
        "extended": True,
        "supplements": {"status":None,"code":code,"name":None,"level":None},
        "searchType": 1,
        "page": page,
        "per_page": 100
    }
    res = requests.post('https://isga.obrnadzor.gov.ru/api/spa/register/getRegisterTable', json=data)
    orgs = res.json()['data']
    return orgs
```

#### Пишем функцию, которая будет получать ИНН нужной нам организации

```python
id = '0B111212-1113-0C12-0D0F-0F0C0F110E12110C120B'
res = requests.get(f'https://isga.obrnadzor.gov.ru/api/spa/register/certificate/{id}/')
```

```python
res.json()
```

```python
res.json()['inn']
```

```python
def get_inn_by_id(id):
  inn = requests.get(f'https://isga.obrnadzor.gov.ru/api/spa/register/certificate/{id}/').json()['inn']
  return inn
```

#### Пишем функцию, которая оставляет только нужные организации (с учетом правовой формы)

```python
for org in orgs_page_1:
  name = org['organization_name'].lower()
  for phrase in ['некоммерческая', 'частная']:
    if phrase in name:
      inn = get_inn_by_id(org['id'])
      print(f'{name}: {inn}')
```

```python
def get_proper_orgs(orgs_lst, phrases_lst):
  proper_orgs_lst = []
  for org in orgs_lst:
    name = org['organization_name'].lower()
    for phrase in phrases_lst:
      if phrase in name:
        inn = get_inn_by_id(org['id'])
        proper_orgs_lst.append(inn)
  return proper_orgs_lst
```

#### **Собираем воедино**

```python
max_pages = get_max_pages("09.02.11")
print(f"max_pages: {max_pages}")

proper_orgs_lst = []
page = 1
while page <= max_pages:
  print(f"page: {page}")
  orgs_lst = get_orgs_list("09.02.11", page)
  new_proper_orgs_lst = get_proper_orgs(orgs_lst, ['некоммерческая', 'частная'])
  print(new_proper_orgs_lst)
  proper_orgs_lst.extend(new_proper_orgs_lst)
  page += 1
  break
```

## Обогатим датасет данными по выручке

#### Пишем функцию получения финансовых данных по ИНН

```python
url = "https://ruinn.ru/api/search"

inn = '2100011426'

r = requests.get(
    url,
    params={"inn": inn},
)

fin = r.json()["bo"][inn].get('2025')
fin.get('2110')
```

```python
def get_revenue_by_inn(inn):
  url = "https://ruinn.ru/api/search"

  try:
    r = requests.get(
        url,
        params={"inn": inn},
    )

    fin = r.json()["bo"][inn].get('2025')
    return fin.get('2110')

  except:
    return None
```

#### Применяем функцию к собранному ранее датасету организаций

```python
proper_orgs_lst
```

```python
revenue_data = {}
for inn in proper_orgs_lst:
  print(inn)
  revenue_data[inn] = get_revenue_by_inn(inn)
```
