```python
class Extraction:
    @classmethod
    def from_csv(cls, filepath):
        sales = []
        with open(filepath, 'r', encoding='utf-8') as f:
            lines = f.readlines()

        for line in lines[1:]:                      # пропускаем заголовок
            line = line.strip()                     # убираем \n
            if not line:
                continue
            id_, date, amount, product = line.split(';')
            sales.append(Sale(int(id_), date, float(amount), product))

        return sales
```

```python
class Sale:

    def __init__(self, id, date, amount, product):
        self.__id = id
        self.__date = date
        self.__amount = amount
        self.__product = product

    @property
    def get_id(self):
        return self.__id

    @get_id.setter
    def get_id(self, value):
        self.__id = value

    @property
    def get_date(self):
        return self.__date

    @get_date.setter
    def get_date(self, value):
        self.__date = value

    @property
    def get_amount(self):
        return self.__amount

    @get_amount.setter
    def get_amount(self, value):
        self.__amount = value

    @property
    def get_product(self):
        return self.__product

    @get_product.setter
    def get_product(self, value):
        self.__product = value
```
