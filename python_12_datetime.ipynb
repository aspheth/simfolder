```python
def most_frequent_day(dates):
  days = []
  for x in dates:
    days.append(datetime.strptime(x,"%Y-%m-%d" ).strftime('%A'))
    res = Counter(days)
  max_val = max(res.values())
  return [x for x,y in res.items() if y == max_val]

```

```python
dates2 = ['2023-05-28 12:00:00', '2023-05-28 13:00:00', '2023-05-28 14:00:00', '2023-05-28 19:00:00']
```

```python
ddays = []
for x in dates2:
  ddays.append(datetime.strptime(x,'%Y-%m-%d %H:%M:%S'))
```

```python
ddays
```

```python
delta = []
for i in range(len(ddays)-1):
  delta.append((ddays[i+1]-ddays[i]).total_seconds())

```
